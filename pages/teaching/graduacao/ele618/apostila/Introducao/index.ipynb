## O Problema do Projetista

Em um sistema elétrico industrial, o engenheiro **não escolhe a carga** — ela é imposta pelo processo produtivo. O projeto sempre parte de três informações dadas:

::: {.incremental}
- as características da **carga mecânica** a ser acionada;
- as **condições ambientais** do local de instalação;
- a **alimentação elétrica** disponível na planta.
:::

. . .

Só depois dessa análise é que se seleciona o motor mais adequado.

## Visão Geral: o Motor como Conversor Eletromecânico

- **Conversor eletromecânico:** Energia elétrica $\leftrightarrow$ Energia mecânica.

## Do Motor à Proteção

Uma vez selecionado o motor, o restante do projeto decorre dele:

::: {.incremental}
1. o motor e seu ponto de operação definem as **correntes de alimentação**;
2. as correntes definem a especificação de **cabos de potência**;
3. cabos e correntes definem o **sistema de proteção**;
4. a demanda energética resultante permite avaliar o **sistema elétrico de potência (SEP)** da planta — transformadores, geradores, subestações.
:::

. . .


. . .

A especificação correta da proteção é crucial para a segurança de equipamentos e pessoas.

## Objetivo do Material

Esta apostila discute a **seleção de motores** e seus métodos de acionamento, com maior atenção ao motor — o elemento de maior complexidade para especificar corretamente.

São abordados:

- a **operação da máquina elétrica** (torque, inércia, transmissão mecânica, dinâmica do acionamento);
- a **seleção da máquina elétrica** (tempo de partida, potência mecânica, grandezas elétricas, parâmetros construtivos).

. . .

::: aside
Todo o conteúdo desta apostila tem como referência principal a obra de LOBOSCO, O. S.; DIAS, J. L. P. C. **Seleção e Aplicação de Motores Elétricos** (Vol. 1 e 2). São Paulo: McGraw-Hill/Siemens, 1988.
:::

# Referências

TOLEDO JÚNIOR, J. A.; MAIA, T. A. C. **Apostila de Sistemas Elétricos Industriais**. Belo Horizonte: Universidade Federal de Minas Gerais (UFMG), 2024. Disponível em: [nextcloud.fahren.com.br/.../Apostila de Sistemas Elétricos Industriais.pdf](https://nextcloud.fahren.com.br/public.php/dav/files/DAKYYJ8p8QpteZW/Apostila%20de%20Sistemas%20El%C3%A9tricos%20Industriais.pdf).

# Exercícios

## Exercício 1

**Qual é a ordem correta de decisões em um projeto de sistema elétrico industrial: (a) especificar proteção, (b) selecionar o motor, (c) caracterizar a carga mecânica, (d) especificar cabos de potência?**

. . .

::: {.incremental}
1. (c) Caracterizar a carga mecânica e o ambiente;
2. (b) Selecionar o motor compatível;
3. (d) Especificar os cabos a partir da corrente do motor;
4. (a) Especificar a proteção a partir dos cabos e correntes.
:::

## Exercício 2

**Por que a carga mecânica não pode ser escolhida pelo projetista, ao contrário do motor?**

. . .

Porque a carga é imposta pelo processo produtivo da planta industrial — é uma característica do equipamento acionado e da aplicação, não um componente comercial que se escolhe de um catálogo. O motor, ao contrário, é selecionado dentre as opções disponíveis no mercado para melhor se adequar àquela carga já definida.

# Obrigado!

<div style="text-align: center;">

[http://thalesmaia.com/](http://thalesmaia.com/)

</div>
